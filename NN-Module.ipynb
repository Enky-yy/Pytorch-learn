```python
import torch
import torch.nn as nn
```

```python
class Model(nn.Module):
    def __init__(self, num_features):
        super().__init__()
        self.linear = nn.Linear(num_features, 1)
        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        out = self.linear(x)
        out = self.sigmoid(out)
        return out

```

```python
features = torch.rand(10,5)
```

```python
model = Model(features.shape[1])
```

```python
pred = model(features)
```

```python
pred
```

```python
pred = (pred > 0.5).float()
```

```python
pred
```

```python
from torchinfo import  summary
```

```python
summary(model, input_size=(10,5))
```

```python
class Model(nn.Module):
    def __init__(self, num_features):
        super().__init__()
        self.linear1 = nn.Linear(num_features, 3)
        self.rulu = nn.ReLU()
        self.linear2 = nn.Linear(3, 1)
        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        out = self.linear1(x)
        out = self.rulu(out)
        out = self.linear2(out)
        out = self.sigmoid(out)
        return out
```

```python
torch.manual_seed(43)
features = torch.rand(10, 5)
model = Model(features.shape[1])
```

```python
pred = model(features)
```

```python
pred
```

```python
summary(model, input_size=(10,5))
```

## Sequential Container

```python
class Model(nn.Module):
    def __init__(self, num_features):
        super().__init__()
        self.cont = nn.Sequential(nn.Linear(num_features, 3),
                                  nn.ReLU(),
                                  nn.Linear(3, 1),
                                  nn.Sigmoid())

    def forward(self, x):
        out = self.cont(x)
        return out
```

```python
torch.manual_seed(43)
features = torch.rand(10, 5)
model = Model(features.shape[1])
```

```python
pred = model(features)
```
