```python
size = 10000
```

```python
matrix_cpu1 = torch.rand(size,size)
matrix_cpu2 = torch.rand(size,size)

start_time = time.time()
matrix_cpu3 = torch.matmul(matrix_cpu1, matrix_cpu2)
cpu_time = time.time() - start_time

print(cpu_time)

matrix_gpu1 = torch.rand(size,size, device=torch.device('cuda'))
matrix_gpu2 = torch.rand(size,size, device=torch.device('cuda'))

start_time = time.time()
matrix_gpu3 = torch.matmul(matrix_gpu1, matrix_gpu2)
gpu_time = time.time() - start_time

print(gpu_time)

print(cpu_time/gpu_time)
```
